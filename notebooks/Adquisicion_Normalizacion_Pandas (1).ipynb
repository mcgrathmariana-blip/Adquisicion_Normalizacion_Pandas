```python
import os

# Crear las carpetas del proyecto
os.makedirs('datos', exist_ok=True)
os.makedirs('salida', exist_ok=True)

print("Estructura de carpetas creada correctamente.")
```

```python
import pandas as pd
import numpy as np
```

```python
transacciones = {
    'id_producto': [101, 102, 103, 101, 104, 105, 102, 103, 104, 105],
    'cantidad': [2, 1, 3, 4, 2, 1, 5, 2, 3, 2],
    'fecha': [
        '2026-01-05', '2026-01-06', '2026-01-07',
        '2026-01-08', '2026-01-09', '2026-01-10',
        '2026-01-11', '2026-01-12', '2026-01-13',
        '2026-01-14'
    ]
}

df_transacciones = pd.DataFrame(transacciones)

display(df_transacciones)
```

```python
productos = {
    'id_producto': [101, 102, 103, 104, 105],
    'nombre_producto': [
        'Laptop',
        'Mouse',
        'Teclado',
        'Monitor',
        'Auriculares'
    ],
    'precio_unitario': [1200, 25, 50, 350, 80]
}

df_productos = pd.DataFrame(productos)

display(df_productos)
```

```python
# Guardar las transacciones en formato CSV
df_transacciones.to_csv(
    'datos/transacciones.csv',
    index=False
)

# Guardar los productos en formato Excel
df_productos.to_excel(
    'datos/productos.xlsx',
    index=False
)

print("Archivos CSV y Excel guardados correctamente.")
```

```python
# Leer el archivo CSV
df_transacciones = pd.read_csv('datos/transacciones.csv')

# Leer el archivo Excel
df_productos = pd.read_excel(
    'datos/productos.xlsx',
    engine='openpyxl'
)

print("Transacciones:")
display(df_transacciones)

print("Productos:")
display(df_productos)
```

```python
# Convertir la fecha al formato datetime
df_transacciones['fecha'] = pd.to_datetime(
    df_transacciones['fecha'],
    errors='coerce'
)

# Convertir los identificadores a valores numéricos
df_transacciones['id_producto'] = pd.to_numeric(
    df_transacciones['id_producto'],
    errors='coerce'
).astype('Int64')

df_productos['id_producto'] = pd.to_numeric(
    df_productos['id_producto'],
    errors='coerce'
).astype('Int64')

# Convertir cantidades y precios a valores numéricos
df_transacciones['cantidad'] = pd.to_numeric(
    df_transacciones['cantidad'],
    errors='coerce'
)

df_productos['precio_unitario'] = pd.to_numeric(
    df_productos['precio_unitario'],
    errors='coerce'
)

# Eliminar registros con valores nulos en campos esenciales
df_transacciones = df_transacciones.dropna(
    subset=['id_producto', 'cantidad', 'fecha']
)

df_productos = df_productos.dropna(
    subset=['id_producto', 'nombre_producto', 'precio_unitario']
)

print("Limpieza y normalización completadas.")

print("\nInformación de transacciones:")
df_transacciones.info()

print("\nInformación de productos:")
df_productos.info()
```

```python
# Integrar las dos fuentes mediante id_producto
df_ventas = pd.merge(
    df_transacciones,
    df_productos,
    on='id_producto',
    how='inner'
)

print("Integración de datos completada.")
print("Dimensiones del dataset:", df_ventas.shape)

display(df_ventas)
```

```python
# Calcular el importe total de cada venta
df_ventas['total_venta'] = (
    df_ventas['cantidad'] * df_ventas['precio_unitario']
)

print("Cálculo de ventas completado.")

display(df_ventas)
```
